
# Checkpoint 01 - Data Visualization
## Spotify User Behavior Analysis



## 1. Dataset Apresentação

Dataset: Spotify User Behavior and Pattern  
Fonte: Kaggle  

Este dataset contém informações sobre o comportamento dos usuários no Spotify, incluindo tempo de uso, preferências musicais e padrões de consumo.



## 2. Perguntas Investigativas

1. Qual é a distribuição do tempo de uso diário?
2. Qual tipo de conteúdo é mais consumido?
3. Existe relação entre tempo de uso e número de sessões?
4. Quais são os gêneros mais populares?


```

import pandas as pd
import plotly.express as px

# Carregar dataset
df = pd.read_csv('spotify_user_behavior.csv')

df.head()
```

```

df = df.dropna()
df.info()
```

## 3. Gráfico 1 - Tempo de uso diário

```

px.histogram(df, x='daily_usage_time', title='Distribuição do Tempo de Uso Diário')
```

## 4. Gráfico 2 - Tipo de conteúdo

```

px.bar(df['content_type'].value_counts().reset_index(),
       x='count', y='content_type',
       title='Tipo de Conteúdo Consumido')
```

## 5. Gráfico 3 - Tempo vs Sessões

```

px.scatter(df, x='daily_usage_time', y='session_per_day',
           title='Tempo de Uso vs Sessões por Dia')
```

## 6. Gráfico 4 - Gêneros

```

px.pie(df, names='favorite_genre', title='Distribuição de Gêneros')
```


## 7. Conclusão

A análise permitiu identificar padrões de uso do Spotify, preferências musicais e comportamento dos usuários.
